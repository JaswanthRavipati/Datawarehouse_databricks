```python
%sql
SELECT DISTINCT c.COUNTRY
FROM workspace.gold.dim_customers AS c
INNER JOIN workspace.gold.fact_sales AS s
ON c.customer_key = s.customer_key
WHERE COUNTRY NOT IN ('n/a');
```

```python
%sql
SELECT p.PRODUCT_ID,
p.PRODUCT_NAME, 
p.CATEGORY,
p.SUBCATEGORY
FROM workspace.gold.dim_products AS p
INNER JOIN workspace.gold.fact_sales AS s
ON p.product_key = s.product_key;
```
